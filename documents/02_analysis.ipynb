##### -> Analizo correlaciones numéricas

```python
df_corr = df_covid.corr(numeric_only=True)
df_corr
```

```python
# Crear una matriz booleana de máscara
mask = np.triu(np.ones_like(df_corr, dtype=bool))

# Crear un heatmap con la máscara
sns.heatmap(df_corr, 
            annot = True, 
            fmt = ".2f", 
            cmap = "mako", 
            vmax = 1, 
            vmin = -1, 
            mask = mask);
```

```python
# Ordenar el DataFrame por año (por si acaso)
df_covid_sorted = df_covid.sort_values('account_creation_year')
```

```python
# Histplot con eje y ordenado por año
sns.histplot(x='account_creation_year', data=df_covid_sorted, discrete=True, kde=True)
plt.title('Años creación cuentas')
plt.xlabel('Año')
plt.ylabel('Cantidad de cuentas')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
top_langs = df_covid['tweet_language'].value_counts().head(10)
sns.barplot(x=top_langs.index, y=top_langs.values, hue= top_langs)
plt.title('Top 10 idiomas de los tweets')
plt.xlabel('Idioma')
plt.ylabel('Cantidad de tweets')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df_covid, x='verified_or_non_verified', y='retweets_received', estimator='mean')
plt.title('Media de retweets según verificación de la cuenta')
plt.xlabel('Cuenta verificada o no')
plt.ylabel('Retweets promedio')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df_covid, x='verified_or_non_verified', y='likes_received', estimator='mean')
plt.title('Media de likes según verificación de la cuenta')
plt.xlabel('Cuenta verificada o no')
plt.ylabel('Likes promedio')
plt.show()
```

```python
top_paises = df_covid['tweet_country'].value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_paises.values, y=top_paises.index)
plt.title('Top 10 países con más tweets')
plt.xlabel('Número de tweets')
plt.ylabel('País')
plt.tight_layout()
plt.show()
```
